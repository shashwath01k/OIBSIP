```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Unemployment Analysis in India

## Objective
To analyze unemployment trends in India and understand the impact of Covid-19 on unemployment rates.

## Technologies Used
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn

```python
df1 = pd.read_csv("Unemployment in India.csv")

df2 = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")
```

```python
df1.head()
```

```python
df2.head()
```

```python
df1.info()
```

```python
df2.info()
```

```python
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

print(df1.columns)
print(df2.columns)
```

```python
print(df1.isnull().sum())
```

```python
print(df2.isnull().sum())
```

```python
df1 = df1.dropna()

df1.info()
```

```python
df1['Date'] = pd.to_datetime(df1['Date'])

df2['Date'] = pd.to_datetime(df2['Date'])
```

```python
df1.describe()
```

```python
df1['Estimated Unemployment Rate (%)'].mean()
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=df2,
    x='Date',
    y='Estimated Unemployment Rate (%)'
)

plt.title("Unemployment Rate During Covid-19")
plt.xticks(rotation=45)

plt.show()
```

```python
state_unemployment = df1.groupby('Region')[
    'Estimated Unemployment Rate (%)'
].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(12,8))

state_unemployment.plot(kind='bar')

plt.title("Average Unemployment Rate by State")
plt.xlabel("State")
plt.ylabel("Unemployment Rate (%)")

plt.show()
```

```python
sns.barplot(
    data=df1,
    x='Area',
    y='Estimated Unemployment Rate (%)'
)

plt.title("Rural vs Urban Unemployment Rate")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df1['Estimated Labour Participation Rate (%)'],
    bins=20
)

plt.title("Labour Participation Rate Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df1.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

# Conclusion

The analysis of unemployment data in India shows a significant increase in unemployment during the Covid-19 period, especially in mid-2020. Different regions experienced different unemployment levels, and noticeable differences were observed between rural and urban areas. Data visualization and statistical analysis helped in understanding unemployment trends and labour participation patterns across India.
